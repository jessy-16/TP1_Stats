## Exercice 1 · Quelle valeur « typique » ?

1. Calculer la moyenne, la médiane et le mode du **prix du billet** (`fare`) et de l'**âge** (`age`).
2. Pour chaque variable, quelle valeur choisiriez-vous pour décrire un passager « typique » ? Justifier.
3. Retirer le 1 % des billets les plus chers. Comment bougent la moyenne et la médiane ?

```python
# 1. Moyenne, médiane, mode de fare et age
titanic["fare"].mean().__round__(2)
titanic["fare"].median().__round__(2)
titanic["fare"].mode()[0]
titanic["fare"].skew()
```

```python
# 3. Sans le 1 % des billets les plus chers
seuil_p99 = titanic["fare"].quantile(0.99)
fare_sans_top = titanic.loc[titanic["fare"] < seuil_p99, "fare"]

print(f"Billets retirés : {len(titanic) - len(fare_sans_top)}")
print(f"Moyenne : {titanic['fare'].mean():.2f} -> {fare_sans_top.mean():.2f}")
print(f"Médiane : {titanic['fare'].median():.2f} -> {fare_sans_top.median():.2f}")
```
